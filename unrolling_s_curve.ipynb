```python
from sklearn.datasets import make_s_curve
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
X, color = make_s_curve(n_samples=1500, random_state=42)
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=color, cmap=plt.cm.Spectral)
ax.view_init(4, -72)
plt.title("The 3D S-Curve (Folded Data)")
plt.show()
```

```python
from sklearn.manifold import Isomap
```

```python
isomap = Isomap(n_neighbors=10, n_components=2)
X_unrolled = isomap.fit_transform(X)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_unrolled[:, 0], X_unrolled[:, 1], c=color, cmap=plt.cm.Spectral)
plt.title("Unrolled S-Curve (2D Projection via Isomap)")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.show()
```
