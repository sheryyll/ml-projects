```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

```python
wine = load_wine()
```

```python
X = wine.data
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
```

```python
cluster_labels = kmeans.fit_predict(X_scaled)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
centroids = pca.transform(kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(9, 6))
```

```python
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis', alpha=0.7, edgecolors='k', s=70)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label='Cluster Centers', edgecolors='white')

plt.title("Unsupervised Learning: K-Means Clustering on Wine Data", fontsize=14)
plt.xlabel("Principal Component 1 (Largest Data Variance)", fontsize=12)
plt.ylabel("Principal Component 2 (Second Largest Variance)", fontsize=12)

plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
