```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.neighbors import LocalOutlierFactor
```

```python
X, _ = make_blobs(n_samples=200, centers=1, cluster_std=1.0, random_state=42)
```

```python
outliers = np.random.uniform(low=-8, high=8, size=(20, 2))
```

```python
X_full = np.vstack([X, outliers])
```

```python
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.1)

predictions = lof.fit_predict(X_full)
```

```python
normal_points = X_full[predictions == 1]
outlier_points = X_full[predictions == -1]

print("Normal points:", len(normal_points))
print("Outliers detected:", len(outlier_points))
```

```python
plt.scatter(normal_points[:, 0], normal_points[:, 1], label="Normal")
plt.scatter(outlier_points[:, 0], outlier_points[:, 1], label="Outliers")

plt.legend()
plt.title("Local Outlier Factor Detection")
plt.show()
```
