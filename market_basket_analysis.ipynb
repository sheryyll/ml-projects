```python
from sklearn.preprocessing import MultiLabelBinarizer
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = [['Milk', 'Bread', 'Eggs'],
           ['Bread', 'Butter'],
           ['Milk', 'Bread', 'Butter'],
           ['Milk', 'Eggs'],
           ['Bread', 'Eggs', 'Butter']]
```

```python
mlb = MultiLabelBinarizer()
sparse_matrix = mlb.fit_transform(dataset)
```

```python
df = pd.DataFrame(sparse_matrix, columns=mlb.classes_)
print("Transaction Matrix:")
print(df)
```

```python
support = df.mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(8, 4))
support.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Item Support (Popularity)")
plt.ylabel("Percentage of Transactions")
plt.show()
```

```python
bread_and_butter = df[(df['Bread'] == 1) & (df['Butter'] == 1)].shape[0]
total_bread = df[df['Bread'] == 1].shape[0]
```

```python
confidence = (bread_and_butter / total_bread) * 100
print(f"Confidence (Bread -> Butter): {confidence:.2f}%")
```
